```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/raw/hasil_survey.csv")

data.head()
```

```python
data = data.drop(columns=['Timestamp'], errors='ignore')

print("Shape:", data.shape)
print("\nKolom:")
print(data.columns.tolist())
```

```python
# 1. Bersihkan spasi di awal/akhir nama kolom CSV Anda
data.columns = data.columns.str.strip()

# 2. Jalankan ulang filter kolom pertanyaan
metadata_cols = [
    "Jenis Kelamin",
    "Semester saat ini",
    "Angkatan",
    "Departemen",
    "Program Studi"
]

wb_cols = [f"WB{i}" for i in range(1, 10)]
acd_cols = [f"ACD{i}" for i in range(1, 9)]
soc_cols = [f"SOC{i}" for i in range(1, 9)]
car_cols = [f"CAR{i}" for i in range(1, 9)]

open_text_cols = [
    "Hal apa yang paling membuat Anda tetap bertahan dan bersemangat menjalani perkuliahan sehari-hari?",
    "Apa kendala akademik yang paling sering Anda alami dan menurut Anda perlu mendapat perhatian kampus?",
    "Apa yang membuat Anda merasa lebih diterima dan nyaman berinteraksi di lingkungan kampus?",
    "Dukungan apa yang paling Anda butuhkan dari kampus untuk meningkatkan kesiapan menghadapi dunia kerja?"
]

question_cols = [
    col for col in data.columns
    if col not in metadata_cols + open_text_cols
]

expected_count = len(wb_cols + acd_cols + soc_cols + car_cols) # Harus 33

print(f"Jumlah pertanyaan terdeteksi: {len(question_cols)}") # Sekarang seharusnya 33

if len(question_cols) != expected_count:
    raise ValueError(
        f"Jumlah pertanyaan Likert terdeteksi {len(question_cols)}, "
        f"seharusnya {expected_count}. Periksa kolom CSV."
    )

new_names = wb_cols + acd_cols + soc_cols + car_cols

data = data.rename(
    columns=dict(zip(question_cols, new_names))
)

print("Berhasil rename kolom!")
```
