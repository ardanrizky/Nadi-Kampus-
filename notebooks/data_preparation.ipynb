```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/raw/hasil_survey.csv")

data.head()
```

```python
print("Shape:", data.shape)
print("\nKolom:")
print(data.columns.tolist())
```

```python
metadata_cols = [
    "Jenis Kelamin",
    "Semester saat ini",
    "Angkatan",
    "Departemen",
    "Program Studi"
]

wb_cols = [f"WB{i}" for i in range(1, 10)]
acd_cols = [f"ACD{i}" for i in range(1, 9)]
soc_cols = [f"SOC{i}" for i in range(1, 9)]
car_cols = [f"CAR{i}" for i in range(1, 9)]

likert_cols = wb_cols + acd_cols + soc_cols + car_cols

print("Metadata:", metadata_cols)
print("Jumlah indikator Likert:", len(likert_cols))
```

```python
metadata = data[metadata_cols].copy()
likert = data[likert_cols].copy()

print("Metadata shape:", metadata.shape)
print("Likert shape:", likert.shape)
```

```python
print("Missing value:")
print(likert.isna().sum())

print("\nTipe data:")
print(likert.dtypes)

print("\nRange nilai:")
print("Minimum:", likert.min().min())
print("Maximum:", likert.max().max())
```

```python
data.to_csv(
    "../data/processed/data_clean.csv",
    index=False
)

print("Data clean berhasil disimpan.")
print("Shape:", data.shape)
```
