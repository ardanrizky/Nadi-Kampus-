```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/processed/data_lpa.csv")

data.head()
```

```python
profile_count = data["Profile"].value_counts().sort_index()

profile_count
```

```python
profile_percentage = (
    data["Profile"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

profile_percentage
```

```python
score_cols = [
    "WB_Score",
    "ACD_Score",
    "SOC_Score",
    "CAR_Score"
]

profile_summary = (
    data.groupby("Profile")[score_cols]
    .mean()
    .round(2)
)

profile_summary
```

```python
profile_distribution = (
    data["Profile"]
    .value_counts()
    .sort_index()
    .to_frame("Jumlah")
)

profile_distribution["Persentase"] = (
    profile_distribution["Jumlah"]
    / len(data)
    * 100
).round(2)

profile_distribution
```

```python
angkatan_profile = pd.crosstab(
    data["Profile"],
    data["Angkatan"],
    normalize="index"
).mul(100).round(2)

angkatan_profile
```

```python
prodi_profile = pd.crosstab(
    data["Profile"],
    data["Program Studi"],
    normalize="index"
).mul(100).round(2)

prodi_profile
```

```python
profile_summary.to_csv(
    "../data/output/cluster_profile.csv"
)

print("Analisis profile berhasil disimpan.")
```
