```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/processed/data_clean.csv")

data.head()
```

```python
wb_cols = [f"WB{i}" for i in range(1, 10)]
acd_cols = [f"ACD{i}" for i in range(1, 9)]
soc_cols = [f"SOC{i}" for i in range(1, 9)]
car_cols = [f"CAR{i}" for i in range(1, 9)]

print("WB  :", wb_cols)
print("ACD :", acd_cols)
print("SOC :", soc_cols)
print("CAR :", car_cols)
```

```python
data["WB_Score"] = data[wb_cols].mean(axis=1)
data["ACD_Score"] = data[acd_cols].mean(axis=1)
data["SOC_Score"] = data[soc_cols].mean(axis=1)
data["CAR_Score"] = data[car_cols].mean(axis=1)

data[
    ["WB_Score", "ACD_Score", "SOC_Score", "CAR_Score"]
].head()
```

```python
score_cols = [
    "WB_Score",
    "ACD_Score",
    "SOC_Score",
    "CAR_Score"
]

print(data[score_cols].describe())
```

```python
data.to_csv(
    "../data/processed/data_scored.csv",
    index=False
)

print("Data scoring berhasil disimpan.")
print("Shape:", data.shape)
```
